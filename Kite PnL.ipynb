```python
import numpy as np
import pandas as pd
```

```python
df= pd.read_csv("orders (1).csv")
```

```python
df
```

```python
# Convert quantity to numeric executed part only
df["Qty."]= df["Qty."].apply(lambda x: float(x.split("/")[0]))
df
```

```python
df = df[(df["Product"]=="MIS") & (df["Status"]=="COMPLETE")]
df
```

```python
df["Turnover"]= df["Qty."] * df["Avg. price"]
df
```

```python
print(df.dtypes)
```

```python
df["Brokerage"]= (df["Turnover"]*0.0003).clip(upper= 20)
df
```

```python
df["STT"]= df.apply(lambda row: 0.00025* row["Turnover"] if row["Type"]== "SELL" else 0, axis =1)
df
```

```python
df["ETC"]= df["Turnover"]* 0.0000325
df
```

```python
df["SEBI"]= df["Turnover"]*0.000001
df
```

```python
df["GST"]= 0.18*(df["Brokerage"] + df["ETC"])
df
```

```python
df["Stamp Charges"]= df.apply(lambda row: 0.00003* row["Turnover"] if row["Type"]== "BUY" else 0, axis= 1)
df
```

```python
df["Total charges"]= (df["Brokerage"]+ df["STT"]+ df["ETC"]+df["SEBI"]+df["GST"]+ df["Stamp Charges"])
df
```

```python
# (Sheet 1) – Trade-wise Charges Table

df_charges= df[["Type","Instrument","Product","Qty.","Avg. price","Status","Turnover","Brokerage","STT","ETC","SEBI","GST","Stamp Charges","Total charges"]]
```

```python
buy_turnover= df_charges[df_charges["Type"] == "BUY"].groupby("Instrument")["Turnover"].sum()
sell_turnover= df_charges[df_charges["Type"] == "SELL"].groupby("Instrument")["Turnover"].sum()
gross_pnl= sell_turnover- buy_turnover

total_charges= df_charges.groupby("Instrument")["Total charges"].sum()
net_pnl= gross_pnl-total_charges

percentage_charges= (total_charges/gross_pnl.abs())*100

Sheet3= pd.DataFrame({
    "Gross PnL" : gross_pnl,
    "Total charges" : total_charges,
    "Net PnL" : net_pnl,
    "% Charges on Gross PnL" : percentage_charges
}).reset_index()
```
